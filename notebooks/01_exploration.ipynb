```python
import duckdb
import plotly.express as px

from demonstrateur.config import DATA_PROCESSED

con = duckdb.connect()
dvf = (DATA_PROCESSED / 'dvf_corse_2024.parquet').as_posix()
```

```python
# Aperçu
con.execute(f"SELECT * FROM '{dvf}' LIMIT 5").df()
```

```python
# Exemple : nombre de ventes et prix médian par commune
df = con.execute(f"""
    SELECT nom_commune,
           count(*)                      AS nb_ventes,
           median(valeur_fonciere)       AS valeur_mediane
    FROM '{dvf}'
    GROUP BY nom_commune
    ORDER BY nb_ventes DESC
    LIMIT 20
""").df()
df
```

```python
px.bar(df, x='nom_commune', y='nb_ventes', title='Ventes 2024 — top 20 communes (brouillon)')
```
